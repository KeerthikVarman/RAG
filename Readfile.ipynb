```python

################################    Data Parasing ######################3

from langchain_community.document_loaders import DirectoryLoader, TextLoader

dir_path = r"C:\Users\keert\OneDrive\Desktop\RAG\data"

loader = DirectoryLoader(
    dir_path,
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)
docs = loader.load()
print(docs)
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

dir_path = r"C:\Users\keert\OneDrive\Desktop\RAG\data\pdf"

loader = DirectoryLoader(
    dir_path,
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=False
)


docs = loader.load()
print(docs)
```
